```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

spark = SparkSession.builder.appName("Q5").getOrCreate()
```

```python
print("q5")
data = [
    (101, "2026-01-01", 250, "Completed"),
    (102, "2026-01-02", 450, None),
    (101, "2026-01-01", 300, "Pending"),
    (103, "2026-01-03", 150, None),
    (102, "2026-01-02", 450, "Completed"),
    (101, "2026-01-01", 500, None),
    (104, "2026-01-04", 140, "Pending")
]
schema = StructType([
    StructField("user_id", IntegerType()),
    StructField("transaction_date", StringType()),
    StructField("amount", IntegerType()),
    StructField("status", StringType())
])
df = spark.createDataFrame(data, schema)
print("Original DataFrame:")
df.show()
df.na.drop().show()
df.na.fill({'status':'Unknown'}).show()
```

```python
print("q6")
from pyspark.sql.functions import count
data = [(city,) for city in
    ["Mumbai"]*120 + ["Delhi"]*150 + ["Pune"]*80 + ["Chennai"]*110 + ["Jaipur"]*50 + ["Kolkata"]*130
]
df_city = spark.createDataFrame(data, ["city"])
print(df_city.count())
result = df_city.groupBy("city").agg(count("*").alias("total_count")).filter("total_count > 100")
result.show()
```

```python
print("q8")
from pyspark.sql.functions import col
data = [
    (101, "jahanvi", 25, "Premium"),
    (102, "anam", 35, "Basic"),
    (103, "simran", 18, "Premium"),
    (104, "palak", 22, "Basic"),
    (105, "sanvi", 30, "Premium"),
    (106, "teja", 17, "Premium"),
    (107, "anushka", 28, "Premium"),
    (108, "kiara", 31, "Premium")
]
df_users = spark.createDataFrame(data, ["user_id","name","age","subscription"])
df_users.show()
fdf = df_users.filter((col("age").between(18, 30)) & (col("subscription")== "Premium"))
fdf.show()
```

```python
print("q10")
from pyspark.sql.functions import col
from pyspark.sql.types import TimestampType
data = [
    (101, "2026-01-01 10:30:00"),
    (102, "2026-01-02 14:45:00"),
    (103, "2026-01-03 09:15:00"),
    (104, "2026-01-04 18:00:00")
]
df_ts = spark.createDataFrame(data, ["user_id", "raw_timestamp"])
df_ts.show()
df_revised = df_ts.withColumn("event_time", col("raw_timestamp").cast(TimestampType())).drop("raw_timestamp")
df_revised.show()
```

```python
print("q12")
from pyspark.sql.functions import col, trim
data = [
    (101, "palal", "palak@gmail.com"),
    (102, "", "jahanvi@gmail.com"),
    (103, "jahanvi", None),
    (104, "anam", "anam@gmail.com"),
    (105, "", None),
    (106, "simran", "simran@gmail.com"),
    (107, "riya", None)
]
df_users = spark.createDataFrame(data, ["user_id", "username", "email"])
print("Original DataFrame:")
df_users.show()
df_cleaned = df_users.filter((col("email").isNotNull()) & (col("username") != ""))
print("After removing invalid rows:")
df_cleaned.show()
```

```python
from pyspark.sql.functions import min, max, avg, col
data = [
    (1, "Store A", 250.0),
    (2, "Store B", 450.0),
    (3, "Store A", 100.0),
    (4, "Store C", 700.0),
    (5, "Store B", 300.0),
    (6, "Store C", 550.0)
]
df_price = spark.createDataFrame(data, ["id", "store_id", "price"])
df_price.agg(
    min("price"),
    max("price"),
    avg("price")).show()
df_price.groupBy("store_id").agg(
    min("price"),
    max("price"),
    avg("price")).show()
```

```python
print("q13")
from pyspark.sql.functions import sum, col

data = [
    (1, "Store A", 250.0),
    (2, "Store B", None),
    (1, "Store A", 250.0),  
    (3, "Store C", 700.0),
    (4, "Store B", 300.0),
    (3, "Store C", 700.0),  
    (5, "Store A", None),
    (6, "Store C", 550.0)]
df_pipeline = spark.createDataFrame(data, ["id", "store_id", "price"])
df_pipeline.show()
result = (
    df_pipeline
    .dropDuplicates()                          
    .na.fill({"price": 0})                     
    .groupBy("store_id")                       
    .agg(sum("price").alias("total_revenue"))  
    .orderBy("store_id"))
result.show()
```
